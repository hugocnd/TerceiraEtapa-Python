```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
path = r"H:\Terceiro Trimestre\Python\Atividade 5 IDH\Tabela4.csv"
```

```python
df = pd.read_csv(path, sep=';', decimal=',', encoding='latin1', header=1)
df = df.dropna(how='all', axis=1)
```

```python
df
```

```python
df_ordenandoem2024 = df.sort_values(by="2024", ascending=False)
print(df_ordenandoem2024[["Estado", "Sigla", "2024"]])
```

--------------

```python
df["Melhora"] = df["2024"] - df["1991"]

estado_maior_melhora = df.loc[df["Melhora"].idxmax()]
print(
    f"Maior melhora: {estado_maior_melhora['Estado']} ({estado_maior_melhora['Sigla']})"
    f"com ganho de +{estado_maior_melhora['Melhora']:.3f} pontos."
)

piorou = df[df["Melhora"] < 0]
if piorou.empty:
    print("Nenhum estado apresentou queda no IDH no periodo de 1991 a 2024.")
else:
    print("Estados com redução no IDH:")
    print(piorou[["Estado", "1991", "2024", "Melhora"]])
```

```python
df_ordenandoem2024 = df.sort_values(by="2024", ascending=False)
print("------------------------------------")
print(df_ordenandoem2024[["Estado", "Sigla", "2024"]])
```

```python
anos = [c for c in df.columns if str(c).isdigit()]
id_vars = [c for c in df.columns if c not in anos and c != "Melhora"]

df_longo = df.melt(
    id_vars=id_vars, value_vars=anos, var_name="Ano", value_name="IDH"
)

df_longo["Ano"] = df_longo["Ano"].astype(int)
df_longo["IDH"] = pd.to_numeric(df_longo["IDH"], errors="coerce")

print(df_longo.head())
```

```python
df_mg = df_longo[df_longo["Sigla"] == "MG"].sort_values("Ano")

fig, ax = plt.subplots(figsize=(8, 5))
ax.plot(
    df_mg["Ano"],
    df_mg["IDH"],
    marker="o",
    color="#d95f02",
    linewidth=2,
    label="Minas Gerais",
)

ax.set_title("Evolução do IDH — Minas Gerais (1991–2024)", fontsize=12)
ax.set_xlabel("Ano")
ax.set_ylabel("IDH")
ax.set_ylim(0.3, 0.9)
ax.grid(True, linestyle="--", alpha=0.6)
ax.legend()

fig.tight_layout()
plt.show()
```

```python
fig, ax = plt.subplots(figsize=(12, 7))

for sigla, grupo in df_longo.groupby("Sigla"):
    grupo = grupo.sort_values("Ano")
    ax.plot(
        grupo["Ano"],
        grupo["IDH"],
        marker="o",
        markersize=3,
        linewidth=1.5,
        label=sigla,
    )

ax.set_title("Evolução do IDH por Estado (1991–2024)", fontsize=14)
ax.set_xlabel("Ano")
ax.set_ylabel("IDH")
ax.set_ylim(0.3, 0.9)
ax.grid(True, linestyle="--", alpha=0.5)
ax.legend(ncol=3, bbox_to_anchor=(1.02, 1), loc="upper left", title="UF")

fig.tight_layout()
plt.show()
```
